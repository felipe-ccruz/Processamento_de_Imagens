```python
img_brg = cv2.imread('400x400.png')
img_rgb = cv2.cvtColor(img_brg, cv2.COLOR_BGR2RGB)
img_grey = cv2.cvtColor(img_brg, cv2.COLOR_BGR2GRAY)

plt.subplot(1,3,1)
plt.imshow(img_brg)
plt.axis('off')
plt.subplot(1,3,2)
plt.imshow(img_rgb)
plt.axis('off')
plt.subplot(1,3,3)
plt.imshow(img_grey, cmap='gray')
plt.axis('off')
plt.show()
```

```python
altura = img_brg.shape[0]
largura = img_brg.shape[1]
print(f'Altura: {altura} e Largura: {largura}')
```

##Operção geomética de Translação

```python
dx = 40
dy = 40

# Criando uma matriz tridimensional para suportar os 3 canais de cores (RGB)
transfer = np.zeros((altura, largura, 3), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    new_x = x + dx
    new_y = y + dy
    if 0 <= new_x < largura and 0 <= new_y < altura:
      # Usando a imagem GREY para exibição correta de cores no matplotlib
      transfer[new_y][new_x] = img_grey[y, x]


plt.figure(figsize=(5,5))
plt.imshow(transfer)
plt.axis('off')
plt.title('Translação')
plt.show()
```

```python
dx = 40
dy = 40

# Criando uma matriz tridimensional para suportar os 3 canais de cores (RGB)
transfer = np.zeros((altura, largura, 3), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    new_x = (x + dx) % largura
    new_y = (y + dy) % altura
    # Usando a imagem RGB para exibição correta de cores no matplotlib
    transfer[new_y][new_x] = img_rgb[y, x]


plt.figure(figsize=(5,5))
plt.imshow(transfer)
plt.axis('off')
plt.title('Translação')
plt.show()
```

##Operação geomética de Escalamento

```python
sx = 1.5
sy = 1.5

nova_largura = int(largura * sx)
nova_altura = int(altura * sy)

scale = np.zeros((nova_altura, nova_largura, 3), dtype=np.uint8)

for x in range(nova_largura):
  for y in range(nova_altura):
    new_x = int(x / sx)
    new_y = int(y / sy)

    if 0 <= new_x < largura and 0 <= new_y < altura:
      # Usando a imagem RGB para exibição correta de cores no matplotlib
      scale[y][x] = img_rgb[new_y][new_x]


print(f'Nova Altura: {nova_altura} e Nova Largura: {nova_largura}')
print(scale.shape)
plt.figure(figsize=(5,5))
plt.imshow(scale)
plt.axis('off')
plt.title('Escalamento')
plt.show()
```

##Operação geomética de Espelhamento



```python
mirror = np.zeros((largura, altura, 3), dtype=np.uint8)

for x in range(largura):
  for y in range(altura):
    new_x = largura - x - 1
    new_y = altura - y - 1

    # Usando a imagem RGB para exibição correta de cores no matplotlib
    mirror[new_y][new_x] = img_rgb[y][x]

plt.figure(figsize=(5,5))
plt.imshow(mirror)
plt.axis('off')
plt.title('Espelhamento')
plt.show()
```
